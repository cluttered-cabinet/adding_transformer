# Creating examples 

There's some stuff I have to think about when it comes to padding. It's simple math, so things are deterministic. 

- Max input length: 6 (i.e. 'xx+xx=')
- Max output length: 3 (i.e. 'xxx', because max input is 99+99 = 198)
- Input is anything from 0 to 99

The dataset generation is pretty simple since I just need to do a for loop over 0 to 99. 

```python
from src.add_tokenizer import AddTokenizer
from src.utils import get_vocab
```

```python
add_vocab = get_vocab()
add_token = AddTokenizer(vocab=add_vocab)
```

```python
add_vocab['<EOS>']
```

```python
class ExGen:
    def __init__(self, tokenizer):
        self.tknzr = tokenizer

    def gen_ex_set(self):
        ex_list = []
        for i in range(0, 100):
            for j in range(0, 100):
                text = ''.join([str(i),'+', str(j), '=', str(i+j)])
                encoded: list[int] = self.tknzr.encode(text)
                encoded.append(self.tknzr.str_to_int['<EOS>'])
                decoded: str = self.tknzr.decode(encoded[:-1])
                assert text == decoded, f"Text doesn't match decoded:\n\ttext:{text}\n\t{decoded}"
                assert self.tknzr.str_to_int["<EOS>"] == encoded[-1], "End of sequence not matched"
                ex_list.append([text, encoded])
        return ex_list
```

```python
ExGen(add_token).gen_ex_set()[0:10]
```
