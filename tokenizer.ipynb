# Creating the tokenizer

Really I only need a couple things for an LLM that adds: 
- digits 0-9
- addition sign (+)
- equals sign (=)
- \<PAD\>
- \<EOS\>

To create the tokenizer, I'll need a vocab that takes all of these, encodes them, decodes them.

A character level encoder is fine. 

```python
tks = list(map(lambda x: str(x), range(0, 10)))
```

```python
tks.extend(["+", "=", "<PAD>", "<EOS>"])
```

```python
add_vocab = {t: i for (i,t) in enumerate(tks)}
```

This works at a high level, but doesn't handle unknown.

```python
class AddTokenizer:
    def __init__(self, vocab: dict):
        # mappings
        self.str_to_int = vocab
        self.int_to_str = {i:j for (j,i) in vocab.items()}
    def encode(self, text: str) -> list[int]:
        ids = [self.str_to_int[i] for i in text]
        return ids
    def decode(self, ids: list[int]) -> str:
        text = "".join([self.int_to_str[i] for i in ids])
        return text
```

```python
test_tokenizer = AddTokenizer(vocab = add_vocab)
```

```python
test_str = "1+1=2"
```

```python
test_tokenizer.encode(test_str)
```

```python
test_tokenizer.decode(test_tokenizer.encode(test_str))
```
